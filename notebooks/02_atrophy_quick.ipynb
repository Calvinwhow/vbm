```python
# Experimental inputs (patients)
experiments_root = '/Volumes/NIMLAB/dl_archive/Atrophy_Howard/01k_Xiao_IGE_Atrophy/BIDS/sub-*/ses-01'
experiments_gm_pattern = 'mwp1*resam*'
experiments_wm_pattern = 'mwp2*resam*'
experiments_csf_pattern = 'mwp3*resam*'

# Output
output_base_dir = '/path/to/where/you/want/the/files'
session='ses-01'

# Config for calculation of atrophy
control_dist = 'ctrl_dist_young_atrophy_only'   # corresponds to the distribution in ./assets
atrophy_only = True                             # Set False to allow for more conservative calculation
```

```python
import subprocess
import sys
# predefined variables
import os 
current_dir = os.path.dirname(os.path.abspath(os.getcwd()))
mask_path = os.path.join(current_dir, 'assets', 'MNI152_T1_2mm_brain_mask.nii')
control_stats_dir = os.path.join(current_dir, 'assets', control_dist)

cmd = [
    sys.executable, f'{current_dir}/scripts/run_z_scoring.py',
    '--experiments-root', experiments_root,
    '--control-stats-dir', control_stats_dir,
    '--experiments-gm-pattern', experiments_gm_pattern,
    '--experiments-wm-pattern', experiments_wm_pattern,
    '--experiments-csf-pattern', experiments_csf_pattern,
    '--output-base-dir', output_base_dir,
    '--session', session,
    '--mask-path', mask_path,
    '--atrophy-only', atrophy_only
]

subprocess.run(cmd, check=True)
```
